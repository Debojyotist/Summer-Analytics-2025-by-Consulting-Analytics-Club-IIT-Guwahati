### 1. What is name of car that has the highest horsepower?  

```python
import pandas as pd
df = pd.read_csv('Cars.csv')
max_hp_car = df.loc[df["horsepower"].astype(float).idxmax()]
print(f"The car with the highest horsepower is: {max_hp_car['name']} with {max_hp_car['horsepower']} HP.")
```

### 2. How many cars have mpg ≥ 35?

```python
cars_with_high_mpg = df[df['mpg'] >= 35]
num_cars = len(cars_with_high_mpg)
print(f"Number of cars with mpg ≥ 35: {num_cars}")
```

### 3. What is the most common origin for cars with horsepower > 100 and weight < 3000?

```python
filtered_cars = df[(df['horsepower'] > 100) & (df['weight'] < 3000)]
most_common_origin = filtered_cars['origin'].mode()[0]
print(f"The most common origin for cars with horsepower > 100 and weight < 3000 is: {most_common_origin}")
```

### 4. What is the mean acceleration of cars from Japan? (rounded to 2 decimals)

```python
japan_cars = df[df['origin'] == 'japan']
mean_acceleration_japan = japan_cars['acceleration'].mean()
rounded_mean_acceleration = round(mean_acceleration_japan, 2)
print(f"The mean acceleration of cars from Japan is: {rounded_mean_acceleration}")
```

### 5. Which year had the highest average mpg?

```python
average_mpg_by_year = df.groupby('model_year')['mpg'].mean()
year_highest_mpg = average_mpg_by_year.idxmax()
highest_avg_mpg_value = average_mpg_by_year.max()
print(f"The year with the highest average mpg is: {year_highest_mpg} with an average of {highest_avg_mpg_value:.2f} mpg")
```

### Find the car (or cars) with the best ratio of horsepower to weight among all cars that also have above-median mpg.

```python
median_mpg = df["mpg"].median()
filtered_cars = df[df["mpg"] > median_mpg]
best_car = filtered_cars.loc[(filtered_cars["horsepower"] / filtered_cars["weight"]).idxmax()]
print(f"Best car: {best_car['name']} with ratio {best_car['horsepower'] / best_car['weight']:.5f}")
```

### Design a multi-line plot using Matplotlib or Seaborn that shows the evolution of average mpg over the years, separately for each origin

```python
avg_mpg = df.groupby(["model_year", "origin"])["mpg"].mean().reset_index()
sns.set_style("whitegrid")
plt.figure(figsize=(10, 6))
sns.lineplot(data=avg_mpg, x="model_year", y="mpg", hue="origin", marker="o")
plt.title("Evolution of Average MPG Over the Years by Origin")
plt.xlabel("Model Year")
plt.ylabel("Average MPG")
plt.legend(title="Origin")
plt.show()
```

### We define a “consistent” car model as one that was produced over multiple years and had very low variation in mpg across those years (standard deviation < 1.0).

Tasks:

Identify car names that appear in more than one model_year.

For each such name, compute the standard deviation of mpg across years.

Return the car(s) with the lowest variation in mpg, among those with at least 2 appearances and std(mpg) < 1.0.

Report the model name(s), number of appearances, and the average mpg.

Bonus:
Sort the result by number of appearances (descending), then mpg (descending).

```python

df = pd.read_csv('Cars.csv')
car_stats = df.groupby('name').agg(
    num_unique_model_years=('model_year', 'nunique'),
    std_mpg=('mpg', 'std'),
    avg_mpg=('mpg', 'mean'),
    num_appearances=('name', 'count')
).reset_index()
consistent_cars = car_stats[
    (car_stats['num_unique_model_years'] >= 2) & # Appeared in at least 2 unique model years
    (car_stats['std_mpg'] < 1.0) # Standard deviation of mpg less than 1.0
]
consistent_cars_sorted = consistent_cars.sort_values(
    by=['num_appearances', 'avg_mpg'],
    ascending=[False, False]
)
print("Consistent Car Models (low variation in mpg across multiple years):")
if not consistent_cars_sorted.empty:
    for index, row in consistent_cars_sorted.iterrows():
        print(f"  Model Name: {row['name']}")
        print(f"  Number of Appearances: {row['num_appearances']}")
        print(f"  Average MPG: {row['avg_mpg']:.2f}")
        print("-" * 30)
else:
    print("No consistent car models found based on the given criteria.")
```
